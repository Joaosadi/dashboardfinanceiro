```python
import requests
import pandas as pd
import sqlite3
import time
database = "brapidados.db"
API_TOKEN = "cdXudtXJ5wv2ALUe3JjozA"
```

```python
def atualizar_banco_brapi(tickers, token=""):
    conn = sqlite3.connect(database)
    
    for ticker in tickers:
        url = f"https://brapi.dev/api/quote/{ticker}"
        params = {"modules": "financialData,defaultKeyStatistics", "token": token}
        headers = {"Authorization": f"Bearer {token}"} if token else {}
        
        resp = requests.get(url, params=params, headers=headers)
        if resp.status_code == 200:
            info = resp.json()["results"][0]
            
            # Criar um resumo limpo dos dados
            dados_empresa = {
                "ticker": ticker,
                "nome": info.get("longName"),
                "preco": info.get("regularMarketPrice"),
                "pl": info.get("defaultKeyStatistics", {}).get("forwardPE"),
                "pb": info.get("defaultKeyStatistics", {}).get("priceToBook"),
                "roe": info.get("financialData", {}).get("returnOnEquity"),
                "margem_liquida": info.get("financialData", {}).get("profitMargins"),
                "marketcap": info.get("marketCap", 0)
            }
            df = pd.DataFrame([dados_empresa]).fillna(0)
            df.to_sql("indicadores_brapi", conn, if_exists="append", index=False)
            print(f"✓ {ticker} inserido no banco.")
            
    conn.close()

tickers = ["PETR4", "VALE3", "MGLU3", "ITUB4"]
atualizar_banco_brapi(tickers)
```

```python
def download_historical_data(ticker):
    url = f"https://brapi.dev/api/v2/stocks/financial-data?symbols={ticker}&mode=history"
    response = requests.request("GET", url, headers = {
      "Authorization": f"Bearer {API_TOKEN}"
    })
    try:
        data = response.json()["results"]
    except:
        print(response.text)
        return
    df = pd.DataFrame(data[0]["data"]).fillna(0)
    df["symbol"] = data[0]["symbol"]
    with sqlite3.connect(database) as conn:
        df.to_sql("indicadores_historicos", conn, if_exists="append", index=False)

def download_all_historical_data():
    with sqlite3.connect(database) as conn:
        ticker_list = pd.read_sql("SELECT DISTINCT symbol FROM lista_tickers", conn)
    for t in ticker_list:
        print(f"Downloading {t}.")
        download_historical_data(t)
        time.sleep(1)
    with sqlite3.connect(database) as conn:
        df = pd.read_sql("SELECT DISTINCT * FROM indicadores_historicos", conn)
    return df
    
# download_all_historical_data()

def download_free_data():
    ticker_list = ["PETR4", "MGLU3", "VALE3", "ITUB4"]
    for t in ticker_list:
        download_historical_data(t)
        time.sleep(1)
    with sqlite3.connect(database) as conn:
        df = pd.read_sql("SELECT DISTINCT * FROM indicadores_historicos", conn)
    return df

df = download_free_data()
```

```python
def download_ticker_list():
    url = "https://brapi.dev/api/v2/tickers"
    response = requests.request("GET", url, headers = {
      "Authorization": f"Bearer {API_TOKEN}"
    })
    data = response.json()["results"]
    result = []
    for s in data:
        cur = {
            "symbol": s["symbol"],
            "name": s["name"],
            "assetType": s["assetType"],
            "sector": s["sector"],
            "logoUrl": s["logoUrl"]
        }
        result.append(cur)
    data = pd.DataFrame(result)
    return data

def update_ticker_table():
    df = download_ticker_list()
    with sqlite3.connect(database) as conn:
        df.to_sql("lista_tickers", conn, if_exists="replace", index=False)
    return df
update_ticker_table()
```
